## Identify buy signals

### Subtask:
Determine the points where the 9-period EMA crosses above the 21-period EMA as potential buy signals.


**Reasoning**:
Create a boolean column 'buy_signal' indicating when the 9-period EMA crosses above the 21-period EMA.



```python
import numpy as np

# Create buy_signal column
qqq_30min_bars['buy_signal'] = np.where(
    (qqq_30min_bars['ema_9'] > qqq_30min_bars['ema_21']) &
    (qqq_30min_bars['ema_9'].shift(1) <= qqq_30min_bars['ema_21'].shift(1)),
    True,
    False
)

# Display rows with buy signals
display(qqq_30min_bars[qqq_30min_bars['buy_signal'] == True])
```

## Implement backtesting logic

### Subtask:
Write code to simulate trades based on the identified buy signals. (For simplicity in this initial backtest, we can assume exiting a position after a fixed period or at a certain profit/loss).


**Reasoning**:
Initialize backtesting variables, iterate through the DataFrame, simulate trades based on buy signals and a fixed holding period, and store trade details.



```python
# 1. Initialize variables
initial_cash = 100000  # Starting cash in USD
cash = initial_cash
position = 0  # Number of QQQ shares held
trades = []  # List to store executed trades
holding_period = 10 # Hold for 10 periods (30-minute intervals)
entry_price = 0
entry_period = -1

# 2. Iterate through the DataFrame
for i in range(len(qqq_30min_bars)):
    # 3. Check for buy signal and no open position
    if qqq_30min_bars['buy_signal'].iloc[i] and position == 0:
        # 4. Simulate buying
        current_price = qqq_30min_bars['close'].iloc[i]
        # Buy a fixed quantity (e.g., enough for $1000)
        buy_quantity = min(int(cash / current_price), 1000 // current_price) # Buy a small fixed dollar amount or what's available
        if buy_quantity > 0:
            cost = buy_quantity * current_price
            cash -= cost
            position += buy_quantity
            entry_price = current_price
            entry_period = i
            trades.append({
                'type': 'buy',
                'symbol': 'QQQ',
                'quantity': buy_quantity,
                'price': current_price,
                'timestamp': qqq_30min_bars.index[i],
                'cash_after_trade': cash,
                'position_after_trade': position
            })

    # 5. Check for exit condition (fixed holding period) and open position
    elif position > 0 and (i - entry_period) >= holding_period:
        # 6. Simulate selling
        current_price = qqq_30min_bars['close'].iloc[i]
        revenue = position * current_price
        cash += revenue
        profit_loss = (current_price - entry_price) * position
        trades.append({
            'type': 'sell',
            'symbol': 'QQQ',
            'quantity': position,
            'price': current_price,
            'timestamp': qqq_30min_bars.index[i],
            'profit_loss': profit_loss,
            'cash_after_trade': cash,
            'position_after_trade': 0
        })
        position = 0
        entry_price = 0
        entry_period = -1 # Reset entry period

# If there's an open position at the end, close it
if position > 0:
    current_price = qqq_30min_bars['close'].iloc[-1]
    revenue = position * current_price
    cash += revenue
    profit_loss = (current_price - entry_price) * position
    trades.append({
        'type': 'sell',
        'symbol': 'QQQ',
        'quantity': position,
        'price': current_price,
        'timestamp': qqq_30min_bars.index[-1],
        'profit_loss': profit_loss,
        'cash_after_trade': cash,
        'position_after_trade': 0
    })
    position = 0


# 7. Store the list of trades
trades_df = pd.DataFrame(trades)

# Display the trades
display(trades_df)

# Display final cash and total profit/loss
print(f"Initial Cash: {initial_cash:.2f}")
print(f"Final Cash: {cash:.2f}")
print(f"Total Profit/Loss: {cash - initial_cash:.2f}")
```

**Reasoning**:
Initialize backtesting variables, iterate through the DataFrame, simulate trades based on filtered buy signals and a fixed holding period, and store trade details.



```python
# 1. Initialize variables
initial_cash = 100000  # Starting cash in USD
cash = initial_cash
position = 0  # Number of QQQ shares held
trades = []  # List to store executed trades
holding_period = 10 # Hold for 10 periods (30-minute intervals)
entry_price = 0
entry_period = -1

# 2. Iterate through the DataFrame
for i in range(len(qqq_30min_bars)):
    # 3. Check for buy signal and no open position
    if qqq_30min_bars['buy_signal_filtered'].iloc[i] and position == 0:
        # 4. Simulate buying
        current_price = qqq_30min_bars['close'].iloc[i]
        # Buy a fixed quantity (e.g., enough for $1000)
        buy_quantity = min(int(cash / current_price), 1000 // current_price) # Buy a small fixed dollar amount or what's available
        if buy_quantity > 0:
            cost = buy_quantity * current_price
            cash -= cost
            position += buy_quantity
            entry_price = current_price
            entry_period = i
            trades.append({
                'type': 'buy',
                'symbol': 'QQQ',
                'quantity': buy_quantity,
                'price': current_price,
                'timestamp': qqq_30min_bars.index[i],
                'cash_after_trade': cash,
                'position_after_trade': position
            })

    # 5. Check for exit condition (fixed holding period) and open position
    elif position > 0 and (i - entry_period) >= holding_period:
        # 6. Simulate selling
        current_price = qqq_30min_bars['close'].iloc[i]
        revenue = position * current_price
        cash += revenue
        profit_loss = (current_price - entry_price) * position
        trades.append({
            'type': 'sell',
            'symbol': 'QQQ',
            'quantity': position,
            'price': current_price,
            'timestamp': qqq_30min_bars.index[i],
            'profit_loss': profit_loss,
            'cash_after_trade': cash,
            'position_after_trade': 0
        })
        position = 0
        entry_price = 0
        entry_period = -1 # Reset entry period

# If there's an open position at the end, close it
if position > 0:
    current_price = qqq_30min_bars['close'].iloc[-1]
    revenue = position * current_price
    cash += revenue
    profit_loss = (current_price - entry_price) * position
    trades.append({
        'type': 'sell',
        'symbol': 'QQQ',
        'quantity': position,
        'price': current_price,
        'timestamp': qqq_30min_bars.index[-1],
        'profit_loss': profit_loss,
        'cash_after_trade': cash,
        'position_after_trade': 0
    })
    position = 0


# 7. Store the list of trades
trades_df = pd.DataFrame(trades)

# Display the trades
display(trades_df)

# Display final cash and total profit/loss
print(f"Initial Cash: {initial_cash:.2f}")
print(f"Final Cash: {cash:.2f}")
print(f"Total Profit/Loss: {cash - initial_cash:.2f}")
```

```python
import alpaca_trade_api as tradeapi
import pandas as pd
import ta
import time
from datetime import datetime, timedelta

# --- Configuration ---
# Make sure to set up your API keys and base URL
# For live trading, use the live trading URL
API_KEY = 'YOUR_ALPACA_API_KEY'
API_SECRET = 'YOUR_ALPACA_SECRET_KEY'
BASE_URL = 'https://api.alpaca.markets' # Live trading URL
SYMBOL = 'QQQ'
TIMEFRAME = '3Min'
BUY_QUANTITY = 50
PROFIT_TARGET_PERCENTAGE_OF_LENGTH = 0.015 # 1.5% of the red candle length


# --- Initialize Alpaca API ---
api = tradeapi.REST(API_KEY, API_SECRET, base_url=BASE_URL)

# --- Global Variables to track the strategy state ---
current_bars = pd.DataFrame()
position_held = False
entry_price = 0
entry_timestamp = None
red_candle_high = None
red_candle_low = None
red_candle_length = None
crossover_found = False # To track if the 10/50 EMA crossover has happened
red_candle_pattern_found = False # To track if the specific red candle after crossover is identified

# --- Function to fetch historical data (useful for initial setup or catching up) ---
def get_historical_data(symbol, timeframe, limit=1000):
    end_date = datetime.now()
    start_date = end_date - timedelta(days=7) # Fetch data for the last 7 days as a buffer
    bars = api.get_bars(symbol, timeframe, start=start_date.strftime('%Y-%m-%d'), end=end_date.strftime('%Y-%m-%d'), limit=limit).df
    return bars.sort_index() # Ensure data is sorted by timestamp

# --- Function to calculate indicators ---
def calculate_indicators(df):
    if not df.empty:
        # Calculate 10-period EMA
        df['ema_10'] = ta.trend.ema_indicator(close=df['close'], window=10)
        # Calculate 50-period EMA
        df['ema_50'] = ta.trend.ema_indicator(close=df['close'], window=50)
        # Drop rows with NaN values created by indicators
        df.dropna(inplace=True)
    return df

# --- Function to check for the EMA crossover ---
def check_crossover(df):
    global crossover_found
    if not df.empty and len(df) >= 2:
        # Check for bullish crossover (10 EMA crosses above 50 EMA)
        if (df['ema_10'].iloc[-1] > df['ema_50'].iloc[-1]) and \
           (df['ema_10'].iloc[-2] <= df['ema_50'].iloc[-2]):
            crossover_found = True
            print(f"Bullish EMA crossover detected at {df.index[-1]}")

# --- Function to find the first red candle after crossover ---
def find_red_candle_after_crossover(df):
    global red_candle_high, red_candle_low, red_candle_length, red_candle_pattern_found
    if crossover_found and not red_candle_pattern_found:
        # Filter for bars after the crossover point
        bars_after_crossover = df[df.index >= df[
             (df['ema_10'] > df['ema_50']) &
             (df['ema_10'].shift(1) <= df['ema_50'].shift(1))
        ].index[0]]

        # Find the first red candle in the filtered data
        red_candles = bars_after_crossover[bars_after_crossover['close'] < bars_after_crossover['open']]

        if not red_candles.empty:
            first_red_candle = red_candles.iloc[0]
            red_candle_high = first_red_candle['high']
            red_candle_low = first_red_candle['low']
            red_candle_length = red_candle_high - red_candle_low
            red_candle_pattern_found = True
            print(f"First red candle after crossover identified at {first_red_candle.name}")
            print(f"Red candle high: {red_candle_high:.2f}, low: {red_candle_low:.2f}, length: {red_candle_length:.2f}")


# --- Function to check for buy signal and place order ---
def check_buy_signal(df):
    global position_held, entry_price, entry_timestamp
    if red_candle_pattern_found and not position_held:
        current_price_high = df['high'].iloc[-1]
        if current_price_high > red_candle_high:
            try:
                # Place a market order to buy
                order = api.submit_order(
                    symbol=SYMBOL,
                    qty=BUY_QUANTITY,
                    side='buy',
                    type='market',
                    time_in_force='gtc' # Good 'Til Canceled
                )
                print(f"Buy order placed for {BUY_QUANTITY} shares of {SYMBOL} at {df.index[-1]}")
                position_held = True
                # Note: Entry price in live trading would be the execution price,
                # which you can get from order updates. For simplicity here,
                # we'll use the red candle high as the theoretical entry price.
                entry_price = red_candle_high
                entry_timestamp = df.index[-1]

            except Exception as e:
                print(f"Error placing buy order: {e}")

# --- Function to check for exit signal and place order ---
def check_sell_signal(df):
    global position_held, red_candle_low, red_candle_length, entry_price
    if position_held:
        current_price_close = df['close'].iloc[-1]
        current_price_high = df['high'].iloc[-1]

        # Calculate profit target price
        profit_target_price = red_candle_high + (red_candle_length * PROFIT_TARGET_PERCENTAGE_OF_LENGTH)

        # Check for Profit Target or Stop Loss
        if current_price_high >= profit_target_price:
            try:
                # Place a market order to sell (Profit Target)
                 order = api.submit_order(
                    symbol=SYMBOL,
                    qty=position_held, # Sell the entire position
                    side='sell',
                    type='market',
                    time_in_force='gtc'
                )
                 print(f"Sell order placed for {position_held} shares of {SYMBOL} at {df.index[-1]} (Profit Target)")
                 position_held = False
                 # Reset pattern tracking after exiting
                 reset_strategy_state()

            except Exception as e:
                print(f"Error placing sell order (Profit Target): {e}")

        elif current_price_close < red_candle_low:
            try:
                # Place a market order to sell (Stop Loss)
                order = api.submit_order(
                    symbol=SYMBOL,
                    qty=position_held, # Sell the entire position
                    side='sell',
                    type='market',
                    time_in_force='gtc'
                )
                print(f"Sell order placed for {position_held} shares of {SYMBOL} at {df.index[-1]} (Closed Below Red Candle Low)")
                position_held = False
                # Reset pattern tracking after exiting
                reset_strategy_state()

            except Exception as e:
                print(f"Error placing sell order (Stop Loss): {e}")

# --- Function to reset strategy state after an exit ---
def reset_strategy_state():
    global crossover_found, red_candle_pattern_found, red_candle_high, red_candle_low, red_candle_length
    crossover_found = False
    red_candle_pattern_found = False
    red_candle_high = None
    red_candle_low = None
    red_candle_length = None
    print("Strategy state reset after trade exit.")


# --- Main trading loop (simplified - you would typically use streaming data) ---
def live_trading_loop():
    global current_bars
    print("Starting simplified live trading loop...")
    # Fetch some initial historical data to prime the indicators
    current_bars = get_historical_data(SYMBOL, TIMEFRAME, limit=200) # Fetch enough data for indicators

    while True:
        # In a real bot, you would consume real-time data from a stream
        # For this simplified example, we'll just wait and fetch the latest bar periodically
        time.sleep(60) # Wait for 60 seconds (adjust based on timeframe)

        try:
            # Fetch the latest bar (you might need to adjust this to get the *completed* latest bar)
            latest_bar_df = api.get_bars(SYMBOL, TIMEFRAME, limit=1).df
            if not latest_bar_df.empty:
                latest_bar = latest_bar_df.iloc[-1]

                # Append the latest bar if it's new
                if latest_bar.name not in current_bars.index:
                    current_bars = pd.concat([current_bars, latest_bar_df])
                    current_bars = current_bars.sort_index().tail(200) # Keep a rolling window of bars

                    # Recalculate indicators on the updated data
                    processed_bars = calculate_indicators(current_bars.copy())

                    if not processed_bars.empty:
                       # Check for EMA crossover
                       if not crossover_found:
                           check_crossover(processed_bars)

                       # Find the red candle after crossover
                       if crossover_found and not red_candle_pattern_found:
                            find_red_candle_after_crossover(processed_bars)

                       # Check for buy signal
                       if red_candle_pattern_found and not position_held:
                           check_buy_signal(processed_bars)

                       # Check for sell signal (if holding a position)
                       if position_held:
                            check_sell_signal(processed_bars)

        except Exception as e:
            print(f"Error in trading loop: {e}")
            # Implement more sophisticated error handling and potentially retry logic
            time.sleep(10) # Wait before retrying

# --- To run the simplified live trading loop (in a real bot, this would be managed differently) ---
# This loop will run indefinitely until interrupted.
# You would typically run this in a dedicated environment or server.
# live_trading_loop()

# print("\nSimplified live trading script outline generated.")
# print("Remember to replace 'YOUR_ALPACA_API_KEY' and 'YOUR_ALPACA_SECRET_KEY' with your actual keys.")
# print("This is a simplified example and needs significant additions for robust live trading.")
```

Write code to iterate through the data starting from the period after the identified red candle, check for the buy signal (price above red candle high), simulate the buy trade, and implement the exit strategy (close below red candle low).

**Reasoning**:
Initialize backtesting variables, find the first red candle after the crossover, iterate through the DataFrame from that point, check for the entry condition (price breaking above the red candle's high), simulate buying 50 shares, and implement the exit condition (price closing below the red candle's low). Store the trade details and display the results.

```python
import pandas as pd
import numpy as np

# 1. Initialize variables
initial_cash = 100000  # Starting cash in USD
cash = initial_cash
position = 0  # Number of QQQ shares held
trades = []  # List to store executed trades
entry_price = 0
entry_timestamp = None
red_candle_high = None
red_candle_low = None
red_candle_found = False
buy_triggered = False
buy_quantity = 50 # Fixed quantity to buy
profit_target_percentage_of_length = 0.015 # 1.5% of the red candle length


# 2. Find the first red candle after the crossover (re-using previous logic)
# This assumes 'crossover_points' and 'qqq_5min_bars' are available from previous steps
first_crossover_timestamp = None
if not crossover_points.empty:
    first_crossover_timestamp = crossover_points.index[0]
    # Find the first red candle after the crossover
    red_candles_after_crossover = qqq_5min_bars[
        (qqq_5min_bars['close'] < qqq_5min_bars['open']) &
        (qqq_5min_bars.index > first_crossover_timestamp)
    ].copy()
    if not red_candles_after_crossover.empty:
        first_red_candle_after_crossover = red_candles_after_crossover.iloc[0]
        red_candle_high = first_red_candle_after_crossover['high']
        red_candle_low = first_red_candle_after_crossover['low']
        red_candle_length = red_candle_high - red_candle_low
        red_candle_found = True
        start_index_for_backtest = qqq_5min_bars.index.get_loc(first_red_candle_after_crossover.name) + 1
        print(f"First red candle after crossover found at: {first_red_candle_after_crossover.name}")
        print(f"Red candle high: {red_candle_high:.2f}, low: {red_candle_low:.2f}, length: {red_candle_length:.2f}")
    else:
        print("No red candles found after the first bullish EMA crossover.")
        start_index_for_backtest = len(qqq_5min_bars) + 1 # Set to a value that won't enter the loop


# 3. Iterate through the DataFrame starting from the period after the red candle
if red_candle_found:
    for i in range(start_index_for_backtest, len(qqq_5min_bars)):
        current_price_close = qqq_5min_bars['close'].iloc[i]
        current_price_high = qqq_5min_bars['high'].iloc[i]
        current_price_low = qqq_5min_bars['low'].iloc[i]

        # Calculate profit target price based on the red candle's high and length
        profit_target_price = red_candle_high + (red_candle_length * profit_target_percentage_of_length)


        # Check for buy signal (price breaks above red candle high)
        if not buy_triggered and current_price_high > red_candle_high:
            # Simulate buying
            # Ensure enough cash to buy the fixed quantity
            cost = buy_quantity * qqq_5min_bars['open'].iloc[i] # Assuming execution at open of next bar
            if cost <= cash:
                 # Adjust price to the breakout price (red_candle_high)
                entry_price = red_candle_high
                # Re-calculate cost based on the entry price
                cost = buy_quantity * entry_price

                cash -= cost
                position += buy_quantity
                entry_timestamp = qqq_5min_bars.index[i]
                buy_triggered = True # Prevent multiple buys for this pattern

                trades.append({
                    'type': 'buy',
                    'symbol': 'QQQ',
                    'quantity': buy_quantity,
                    'price': entry_price,
                    'timestamp': qqq_5min_bars.index[i],
                    'cash_after_trade': cash,
                    'position_after_trade': position
                })
                print(f"Buy triggered at {entry_timestamp} at price {entry_price:.2f}")


        # Check for exit condition (profit target or close below red candle low) if holding a position
        if position > 0:
            # Check for Profit Target
            if current_price_high >= profit_target_price:
                 # Simulate selling at the profit target price
                sell_price = profit_target_price
                revenue = position * sell_price
                cash += revenue
                profit_loss = (sell_price - entry_price) * position
                trades.append({
                    'type': 'sell',
                    'symbol': 'QQQ',
                    'quantity': position,
                    'price': sell_price,
                    'timestamp': qqq_5min_bars.index[i], # Log the timestamp when target was hit
                    'profit_loss': profit_loss,
                    'cash_after_trade': cash,
                    'position_after_trade': 0,
                    'exit_reason': 'Profit Target'
                })
                position = 0 # Exit the position
                buy_triggered = False # Allow buying again
                print(f"Sell triggered at {qqq_5min_bars.index[i]} at price {sell_price:.2f} (Profit Target)")

            # Check for exit below red candle low
            elif current_price_close < red_candle_low:
                # Simulate selling
                revenue = position * current_price_close
                cash += revenue
                profit_loss = (current_price_close - entry_price) * position
                trades.append({
                    'type': 'sell',
                    'symbol': 'QQQ',
                    'quantity': position,
                    'price': current_price_close,
                    'timestamp': qqq_5min_bars.index[i],
                    'profit_loss': profit_loss,
                    'cash_after_trade': cash,
                    'position_after_trade': 0,
                    'exit_reason': 'Closed Below Red Candle Low'
                })
                position = 0 # Exit the position
                buy_triggered = False # Allow buying again after exiting a position triggered by this pattern
                print(f"Sell triggered at {qqq_5min_bars.index[i]} at price {current_price_close:.2f} (Closed Below Red Candle Low)")


# If there's an open position at the end, close it at the last price
if position > 0:
    current_price_close = qqq_5min_bars['close'].iloc[-1]
    revenue = position * current_price_close
    cash += revenue
    profit_loss = (current_price_close - entry_price) * position
    trades.append({
        'type': 'sell',
        'symbol': 'QQQ',
        'quantity': position,
        'price': current_price_close,
        'timestamp': qqq_5min_bars.index[-1],
        'profit_loss': profit_loss,
        'cash_after_trade': cash,
        'position_after_trade': 0,
        'exit_reason': 'End of Backtest'
    })
    position = 0


# 4. Store the list of trades
trades_df = pd.DataFrame(trades)

# Display the trades
display(trades_df)

# Display final cash and total profit/loss
print(f"Initial Cash: {initial_cash:.2f}")
print(f"Final Cash: {cash:.2f}")
print(f"Total Profit/Loss: {cash - initial_cash:.2f}")
```

```python
import alpaca_trade_api as tradeapi
import pandas as pd
import ta
import time
from datetime import datetime, timedelta

# --- Configuration ---
# Make sure to set up your API keys and base URL
# For live trading, use the live trading URL
API_KEY = 'PKAZJ2CHZ409JXQEE3JI'
API_SECRET = 'DbmnlcNEM7sTfVG66Q4YfDhlKvfdaHO64evp0dBj'
BASE_URL = 'https://api.alpaca.markets' # Live trading URL
SYMBOL = 'QQQ'
TIMEFRAME = '3Min'
BUY_QUANTITY = 50
PROFIT_TARGET_PERCENTAGE_OF_LENGTH = 0.015 # 1.5% of the red candle length


# --- Initialize Alpaca API ---
api = tradeapi.REST(API_KEY, API_SECRET, base_url=BASE_URL)

# --- Global Variables to track the strategy state ---
current_bars = pd.DataFrame()
position_held = False
entry_price = 0
entry_timestamp = None
red_candle_high = None
red_candle_low = None
red_candle_length = None
crossover_found = False # To track if the 10/50 EMA crossover has happened
red_candle_pattern_found = False # To track if the specific red candle after crossover is identified

# --- Function to fetch historical data (useful for initial setup or catching up) ---
def get_historical_data(symbol, timeframe, limit=1000):
    end_date = datetime.now()
    start_date = end_date - timedelta(days=7) # Fetch data for the last 7 days as a buffer
    bars = api.get_bars(symbol, timeframe, start=start_date.strftime('%Y-%m-%d'), end=end_date.strftime('%Y-%m-%d'), limit=limit).df
    return bars.sort_index() # Ensure data is sorted by timestamp

# --- Function to calculate indicators ---
def calculate_indicators(df):
    if not df.empty:
        # Calculate 10-period EMA
        df['ema_10'] = ta.trend.ema_indicator(close=df['close'], window=10)
        # Calculate 50-period EMA
        df['ema_50'] = ta.trend.ema_indicator(close=df['close'], window=50)
        # Drop rows with NaN values created by indicators
        df.dropna(inplace=True)
    return df

# --- Function to check for the EMA crossover ---
def check_crossover(df):
    global crossover_found
    if not df.empty and len(df) >= 2:
        # Check for bullish crossover (10 EMA crosses above 50 EMA)
        if (df['ema_10'].iloc[-1] > df['ema_50'].iloc[-1]) and \
           (df['ema_10'].iloc[-2] <= df['ema_50'].iloc[-2]):
            crossover_found = True
            print(f"Bullish EMA crossover detected at {df.index[-1]}")

# --- Function to find the first red candle after crossover ---
def find_red_candle_after_crossover(df):
    global red_candle_high, red_candle_low, red_candle_length, red_candle_pattern_found
    if crossover_found and not red_candle_pattern_found:
        # Filter for bars after the crossover point
        bars_after_crossover = df[df.index >= df[
             (df['ema_10'] > df['ema_50']) &
             (df['ema_10'].shift(1) <= df['ema_50'].shift(1))
        ].index[0]]

        # Find the first red candle in the filtered data
        red_candles = bars_after_crossover[bars_after_crossover['close'] < bars_after_crossover['open']]

        if not red_candles.empty:
            first_red_candle = red_candles.iloc[0]
            red_candle_high = first_red_candle['high']
            red_candle_low = first_red_candle['low']
            red_candle_length = red_candle_high - red_candle_low
            red_candle_pattern_found = True
            print(f"First red candle after crossover identified at {first_red_candle.name}")
            print(f"Red candle high: {red_candle_high:.2f}, low: {red_candle_low:.2f}, length: {red_candle_length:.2f}")


# --- Function to check for buy signal and place order ---
def check_buy_signal(df):
    global position_held, entry_price, entry_timestamp
    if red_candle_pattern_found and not position_held:
        current_price_high = df['high'].iloc[-1]
        if current_price_high > red_candle_high:
            try:
                # Place a market order to buy
                order = api.submit_order(
                    symbol=SYMBOL,
                    qty=BUY_QUANTITY,
                    side='buy',
                    type='market',
                    time_in_force='gtc' # Good 'Til Canceled
                )
                print(f"Buy order placed for {BUY_QUANTITY} shares of {SYMBOL} at {df.index[-1]}")
                position_held = True
                # Note: Entry price in live trading would be the execution price,
                # which you can get from order updates. For simplicity here,
                # we'll use the red candle high as the theoretical entry price.
                entry_price = red_candle_high
                entry_timestamp = df.index[-1]

            except Exception as e:
                print(f"Error placing buy order: {e}")

# --- Function to check for exit signal and place order ---
def check_sell_signal(df):
    global position_held, red_candle_low, red_candle_length, entry_price
    if position_held:
        current_price_close = df['close'].iloc[-1]
        current_price_high = df['high'].iloc[-1]

        # Calculate profit target price
        profit_target_price = red_candle_high + (red_candle_length * PROFIT_TARGET_PERCENTAGE_OF_LENGTH)

        # Check for Profit Target or Stop Loss
        if current_price_high >= profit_target_price:
            try:
                # Place a market order to sell (Profit Target)
                 order = api.submit_order(
                    symbol=SYMBOL,
                    qty=position_held, # Sell the entire position
                    side='sell',
                    type='market',
                    time_in_force='gtc'
                )
                 print(f"Sell order placed for {position_held} shares of {SYMBOL} at {df.index[-1]} (Profit Target)")
                 position_held = False
                 # Reset pattern tracking after exiting
                 reset_strategy_state()

            except Exception as e:
                print(f"Error placing sell order (Profit Target): {e}")

        elif current_price_close < red_candle_low:
            try:
                # Place a market order to sell (Stop Loss)
                order = api.submit_order(
                    symbol=SYMBOL,
                    qty=position_held, # Sell the entire position
                    side='sell',
                    type='market',
                    time_in_force='gtc'
                )
                print(f"Sell order placed for {position_held} shares of {SYMBOL} at {df.index[-1]} (Closed Below Red Candle Low)")
                position_held = False
                # Reset pattern tracking after exiting
                reset_strategy_state()

            except Exception as e:
                print(f"Error placing sell order (Stop Loss): {e}")

# --- Function to reset strategy state after an exit ---
def reset_strategy_state():
    global crossover_found, red_candle_pattern_found, red_candle_high, red_candle_low, red_candle_length
    crossover_found = False
    red_candle_pattern_found = False
    red_candle_high = None
    red_candle_low = None
    red_candle_length = None
    print("Strategy state reset after trade exit.")


# --- Main trading loop (simplified - you would typically use streaming data) ---
def live_trading_loop():
    global current_bars
    print("Starting simplified live trading loop...")
    # Fetch some initial historical data to prime the indicators
    current_bars = get_historical_data(SYMBOL, TIMEFRAME, limit=200) # Fetch enough data for indicators

    while True:
        # In a real bot, you would consume real-time data from a stream
        # For this simplified example, we'll just wait and fetch the latest bar periodically
        time.sleep(60) # Wait for 60 seconds (adjust based on timeframe)

        try:
            # Fetch the latest bar (you might need to adjust this to get the *completed* latest bar)
            latest_bar_df = api.get_bars(SYMBOL, TIMEFRAME, limit=1).df
            if not latest_bar_df.empty:
                latest_bar = latest_bar_df.iloc[-1]

                # Append the latest bar if it's new
                if latest_bar.name not in current_bars.index:
                    current_bars = pd.concat([current_bars, latest_bar_df])
                    current_bars = current_bars.sort_index().tail(200) # Keep a rolling window of bars

                    # Recalculate indicators on the updated data
                    processed_bars = calculate_indicators(current_bars.copy())

                    if not processed_bars.empty:
                       # Check for EMA crossover
                       if not crossover_found:
                           check_crossover(processed_bars)

                       # Find the red candle after crossover
                       if crossover_found and not red_candle_pattern_found:
                            find_red_candle_after_crossover(processed_bars)

                       # Check for buy signal
                       if red_candle_pattern_found and not position_held:
                           check_buy_signal(processed_bars)

                       # Check for sell signal (if holding a position)
                       if position_held:
                            check_sell_signal(processed_bars)

        except Exception as e:
            print(f"Error in trading loop: {e}")
            # Implement more sophisticated error handling and potentially retry logic
            time.sleep(10) # Wait before retrying

# --- To run the simplified live trading loop (in a real bot, this would be managed differently) ---
# This loop will run indefinitely until interrupted.
# You would typically run this in a dedicated environment or server.
# live_trading_loop()

# print("\nSimplified live trading script outline generated.")
# print("Remember to replace 'YOUR_ALPACA_API_KEY' and 'YOUR_ALPACA_SECRET_KEY' with your actual keys.")
# print("This is a simplified example and needs significant additions for robust live trading.")
```
